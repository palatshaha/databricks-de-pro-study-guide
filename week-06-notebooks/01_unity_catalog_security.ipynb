# Unity Catalog Security

Explore hierarchical permissions, ownership vs MANAGE, and group-based grants.

```
-- Grant usage hierarchy example
GRANT USAGE ON CATALOG main TO `data_analysts`;
GRANT USAGE ON SCHEMA main.sales TO `data_analysts`;
GRANT SELECT ON TABLE main.sales.orders TO `data_analysts`;
```

```
-- Transfer ownership example
ALTER TABLE main.sales.orders OWNER TO `data_platform_team`;
```
