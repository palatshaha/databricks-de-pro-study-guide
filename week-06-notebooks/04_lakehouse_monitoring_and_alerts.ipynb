# Lakehouse Monitoring and Observability

Query system tables and implement freshness checks.

```
-- Query system audit logs
SELECT * FROM system.access.audit LIMIT 10;
```

```
-- Example freshness query
SELECT table_name,
       DATEDIFF(HOUR, max_event_time, current_timestamp()) AS hours_stale
FROM system.lakehouse_monitoring.metrics
WHERE hours_stale > 4;
```
